This module serves as the main entry point for the CLI-based student marks management system,
now enhanced with data visualization capabilities. It includes user interaction through a command-line menu,
handling of student marks data, and the generation of a bar chart to visually represent student performance.

Key Features:
- Display and manage a menu for user interactions.
- Collect and manage student marks data.
- Utilize the 'fileOperation' module for reading and writing student data to a file.
- Generate and display a bar chart of student marks using Matplotlib for data visualization.

Functions:
- display_menu(): Displays the main menu options to the user.
- get_mark(): Captures a student's name and mark for storage.
- display_mark(): Shows all the stored student marks.
- display_bar_chart(studentMarks): Generates and displays a bar chart representing student marks.
- CLI_menu(): The core function that runs the Command Line Interface menu loop.

The module demonstrates a step forward in modular programming in Python, encapsulating
data visualization alongside basic function definitions, loops, conditionals, and file handling.

When executed as the main program, it initializes the student marks list from a file and
starts the CLI menu loop for user interaction.





```python
#Example Code V4
#Last Updated: 18/09/24
#Author: Firat and Stephen

# Importing the fileOperation module
import fileOperation as FO
import matplotlib.pyplot as plt
from IPython.display import clear_output

# -----------------------------
# Function Definitions
# -----------------------------

def draw_line(length=20, symbol="="):
    """
    Draws a line consisting of a repeated symbol.

    Parameters:
    length (int): The length of the line. Default is 20.
    symbol (str): The symbol used to draw the line. Default is "=".

    Returns:
    None: This function prints the line and returns nothing.
    """
    line = length * symbol
    print(line)

def display_message(size=10, msg=""):
    """
    Displays a message within a frame of symbols.

    Parameters:
    size (int): The total width of the message display including the frame. Default is 10.
    msg (str): The message to be displayed. Default is an empty string.

    Returns:
    None: This function prints the message within a frame and returns nothing.
    """
    symbol = "*"
    size_msg = len(msg)
    side_size = int((size - size_msg) / 2)
    side_line = side_size * symbol
    draw_line(size)
    print(side_line + msg + side_line)
    draw_line(size)
    
def display_mark():
    """
    Displays all stored student marks.

    Parameters:
    None

    Returns:
    None: This function prints each student's name and mark and returns nothing.
    """
    for smark in studentMarks:
        print(smark)


def get_mark():
    """
    Prompts the user to enter a student's name and mark, then stores them.

    Parameters:
    None

    Returns:
    None: This function adds the student's name and mark to a list and returns nothing.
    """
    sName = input("Please enter a student's name: ")
    sMark = input("Please enter the student's mark: ")
    studentMarks.append([sName, sMark])
    print(studentMarks)
# -----------------------------
# Visualisation functions
# -----------------------------

def display_bar_chart(studentMarks):
    """
    Displays a bar chart of student marks.

    Parameters:
    studentMarks (list): A list of student marks where each mark is a [name, score] pair.

    Returns:
    None: This function generates and displays a bar chart.
    """
    # Initialize empty lists for names and scores
    names = []
    scores = []

    # Iterate over studentMarks to populate names and scores lists
    for mark in studentMarks:
        names.append(mark[0])       # Append the student's name
        scores.append(int(mark[1])) # Append the student's score, converting it to an integer

    # Generate and display the bar chart
    plt.bar(names, scores)
    plt.xlabel('Students')
    plt.ylabel('Marks')
    plt.title('Bar Chart of Student Marks')
    plt.show()

# -----------------------------
# Menu Operations
# -----------------------------

def display_menu():
    """
    Displays the main menu options to the user, including options to manage student marks and file operations.

    Parameters:
    None

    Returns:
    None: This function prints the menu options and returns nothing.
    """
    display_message(20, "Menu")
    print("m) get student mark")
    print("d) display marks")
    print("l) load the marks from a file")
    print("s) save the marks to a file")
    print("c) display bar chart of marks")
    print("e) exit")

def CLI_menu():
    """
    Runs the Command Line Interface (CLI) menu loop,
    processing user input for various operations including file handling.

    Parameters:
    None

    Returns:
    None: This function calls other functions based on user input
    and manages the overall program flow.
    """
    global studentMarks
    
    while True:     
        display_menu()
        option = input("Please enter your option: ")
        clear_output()
        if option == "m":
            get_mark()
        elif option == "d":
            display_mark()
        elif option == "c":
            display_bar_chart(studentMarks)
        elif option == "l":
            studentMarks = FO.load_marks()
        elif option == "s":
            FO.save_marks(studentMarks)
        elif option == "e":
            print("Bye bye, end of the program")
            break
        else:
            print("Wrong option")
        input("Please enter any key to see the menu")
        clear_output()
# -----------------------------
# Main Program
# -----------------------------

studentMarks = FO.load_marks()
CLI_menu()
```
