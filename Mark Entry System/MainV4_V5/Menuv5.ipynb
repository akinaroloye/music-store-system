# Interactive Student Marks Management System - Version 5

## Overview

Welcome to Version 5 of the Student Marks Management System. In this iteration, we've transformed the Python script into an interactive Jupyter Notebook experience using `ipywidgets`. This notebook integrates a user-friendly interface with buttons for each functionality, making it easier and more intuitive to manage student marks.

## Functionalities

The system includes the following features, each accessible through a dedicated button:


1. **Get Student Mark:** Input fields to enter a student's name and mark. Submitted data is added to the student marks list.
2. **Display Marks:** Shows the current list of all student marks.
3. **Load Marks:** Loads student marks from a file.
4. **Save Marks:** Saves current student marks to a file.
5. **Display Bar Chart:** Visualizes student marks in a bar chart format.

## Instructions

- Click on the respective buttons to perform each operation.
- For the "Get Student Mark" function, enter the student's name and mark in the provided fields and click 'Submit Mark' to add the data.
- The output for each action will be displayed below the buttons.
- To see a visual representation of the marks, use the "Display Bar Chart" option.

Enjoy the interactive experience of managing student marks with ease and efficiency!




```python
#Example Code V5
#Last Updated: 18/09/24
#Author: Firat and Stephen

import fileOperation as FO
import matplotlib.pyplot as plt
from IPython.display import clear_output
import ipywidgets as widgets
    
# -----------------------------
#    Visualisation functions
# -----------------------------

def display_bar_chart(studentMarks):
    """
    Displays a bar chart of student marks.

    Parameters:
    studentMarks (list): A list of student marks where each mark is a [name, score] pair.

    Returns:
    None: This function generates and displays a bar chart.
    """
    # Initialize empty lists for names and scores
    names , scores = [] , []

    # Iterate over studentMarks to populate names and scores lists
    for mark in studentMarks:
        names.append(mark[0])       # Append the student's name
        scores.append(int(mark[1])) # Append the student's score, converting it to an integer

    # Generate and display the bar chart
    plt.bar(names, scores)
    plt.xlabel('Students')
    plt.ylabel('Marks')
    plt.title('Bar Chart of Student Marks')
    plt.show()
#----------------------------
#Student Matks initialisation    
#----------------------------
studentMarks = FO.load_marks()
```

```python
#---------------------------------
# GUI Design 
#--------------------------------

# Define button click event handlers
def submit_student_mark(b):
    with output:
        clear_output()
        student_name = name_input.value
        student_mark = mark_input.value
        studentMarks.append([student_name, student_mark])
        print(f"Mark added: {student_name}, {student_mark}")

def on_get_mark_clicked(b):
    # Displaying the widgets
    input_form = widgets.VBox([name_input, mark_input, submit_button])
    with output:
        clear_output()
        display(input_form)
    
def on_display_marks_clicked(b):
    with output:
        clear_output()
        for smark in studentMarks:
            print(smark)

def on_load_clicked(b):
    global studentMarks
    with output:
        clear_output()
        #call the functionality of load_marks from fileOperation
        print("Load Marks Functionality")
        studentMarks = FO.load_marks()
        
def on_save_clicked(b):
    with output:
        clear_output()
        # call the functionality of save_marks from fileOperation
        print("Save Marks Functionality")
        FO.save_marks(studentMarks)
def on_bar_chart_clicked(b):
    with output:
        clear_output()
        # call display_bar_chart
        print("Display Bar Chart Functionality")
        display_bar_chart(studentMarks)

# Create buttons for each menu option
btn_get_mark = widgets.Button(description="Get Student Mark")
btn_display_marks = widgets.Button(description="Display Marks")
btn_load = widgets.Button(description="Load Marks")
btn_save = widgets.Button(description="Save Marks")
btn_bar_chart = widgets.Button(description="Display Bar Chart")

# Creating widgets for student name and mark input
name_input = widgets.Text(description="Name:")
mark_input = widgets.Text(description="Mark:")
# Creating a submit button
submit_button = widgets.Button(description="Submit Mark")
# Link the button to its event handler
submit_button.on_click(submit_student_mark)

# Create an output widget
output = widgets.Output()

# Link menu buttons to their event handlers
btn_get_mark.on_click(on_get_mark_clicked)
btn_display_marks.on_click(on_display_marks_clicked)
btn_load.on_click(on_load_clicked)
btn_save.on_click(on_save_clicked)
btn_bar_chart.on_click(on_bar_chart_clicked)


buttons = widgets.HBox([btn_get_mark, btn_display_marks, btn_load, btn_save, btn_bar_chart])

display(buttons, output)
```
